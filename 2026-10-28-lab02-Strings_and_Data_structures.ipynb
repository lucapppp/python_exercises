## Exercise 04
Write a Python snippet that takes a dictionary in input in the form `{index:text}`, where `index` is an integer, while `text` can include any sentence. The code should produce a dictionary where all the items containing a value (as a substring) of another key are removed.

Input: `in_dict={1: "friends", 2: "he is my friend", 3: "harry", 4: "harry potter", 5: "we are friends"}`

Output: `{2: 'he is my friend', 4: 'harry potter', 5: 'we are friends'}`

````python
my_dict = {1: "friends", 2: "he is my friend", 3: "harry", 4: "harry potter", 5: "we are friends"}
new_dict = {}

for key, value in my_dict.items():
    should_i_keep_this = True

    for key2, value2 in my_dict.items():
        if key != key2 and len(value2) <=len(value) and value2 in value:
            should_i_keep_this = False
            break
        
        if should_i_keep_this:
            new_dict[key] = value
print(new_dict)
````
