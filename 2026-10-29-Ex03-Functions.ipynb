## Exercise 00
Write a Python program that computes the average values of the numbers contained in each tuple contained in a list.

Input: nums = [(1,2,3), (4,5,6,7), (24,18,19)]

Output: [2.0, 5.5, 20.333333]

````python
# Solution with list comprehension
nums = [(1,2,3),(4, 5, 6, 7),(24, 18, 19)]
````

````python
# Writing a function
def average_function(input_tuple):
    sum_of_numbers = 0
    for num in input_tuple:
        sum_of_numbers += num
    average = sum_of_numbers/len(input_tuple)
    return average
````

## Exercise 02
Write the function **listify** that:
- takes a function that works on a single input (e.g. abs)
- returns a function that works on a list, which behaviour is identical to apply the input function itself to each element of the list

### Alert: advanced concept!
**This pattern is called *decorator*.**

A decorator is a pattern in Python that allows a developer to add new functionality to an existing object, like a function, without modifying its structure.

A decorator can be called by preponing @name_of_decorator_function to the object definition.
For instance:
```python

def my_decorator(function):

    def inner(input_arg):
        # Do something cool here to add functionality
        return function(input_arg)

    return inner

@my_decorator
def a_normal_function(val)
    return val

# Now, when I apply a_normal_function, the function also does what has been specified in the inner function of my_decorator
```
