# functions
functions are an important topic to learn how to incapsulate some code
there is a problem with what we used till now. This approach works fine for small programs. Unfortunately, it gets confusing and
disorganized very quickly, as the program increases in size and complexity.
We need mechanisms to structure the code, to write code that is:
· Well organized
· Reusable
· Coherent
Typically long code isn't a good idea.
More functionalities is a measure of the goodness of the software
To write less code, while having more functionalities, we need to use
some software architecture methods: functions are one of such methods
Functions help programmers to achieve decomposition and abstraction

Abstraction
Abstraction: representing essential features while hiding internal details
and implementation.
For example, we know what the print()function does, but we do not
know how it works.
It allows to use functionalities without the need of knowing or
understanding the background details.
It is a key concept to reduce the complexity of large systems.
The same function can have more than one implementation (typically the
implementation of low level functions depends on the hardware and
operating system); these details are hidden to the user of the function.

Decomposition
Decomposition: breaking a complexproblem into smaller pieces.
The solution of a large problem is designed as the composition of the solutions of smaller sub-problems.
Smaller pieces are easier to design, implement and maintain.
Decomposition helps reusability: the solutions of smaller problems are easier to be reused.

To make abstraction and decomposition in the same model, i organize the code in modules, which implements sub-task, self-contained, can be reused and keep the code organized
Those modules are:
modules
classes
functions: those are named or unnamed

Functions
We… already know how to use functions!
· Input and output functions: input(…), print(…)
· Type casting functions: int(…), float(…), str(…)
· Utility functions: len(…), min(…), round(…), …
· Methods: dictionary.keys(), list.append(…), set.add(…)
We know how to:
· Run functions;
· Pass parameters (=arguments);
· Get their result.
In order to use a function we just need to know:
· What input it takes (the parameters or arguments of the functions)
· What output it returns
· What it does
It is not necessary to know any implementation detail:
· From the user point of view, functions are black boxes that transform inputs into outputs
Functions:
· Enforce abstraction: all the implementation details can be ignored
when using a function
· Allow to achieve decomposition: each subtask can be implemented in
a function. We can next think of the solution in terms of composition
of functions and reuse the same function many times

Functions can be seen as mini-programs.
When a program starts, it creates a new scope (= context/environment)
When a function in executed, the Python interpreter logically:
 · Creates a new scope just to execute the function; this scope is totally unrelated with the one of the main program and has new variables.
 · Binds the formal parameter to the actual parameters; the value in “x” is bound to the parameter “n” in the function definition;
 · Executes the function in the new context/environment;
 · Takes the value returned by the function and binds it to the output

Python allows function arguments to have default values.
In this way, if the function is called without the argument, the argument
variable is assigned the default value.

```python
def greet(name, surname="", title=""): #se dico così significa che o gli dò valore io o valgono 0
    print( "Hello {} {} {}".format(title, name, surname))
greet("Barack") # Hello Barack
greet("Barack", surname= "Obama") # Hello Barack Obama
greet("Barack", title= "Mr.") # Hello Mr. Barack
greet("Barack", title= "Mr.", surname= "Obama") # Hello Mr. Barack Obama
```

```python
def sum_all(n, *ns):
    res = n
    for elem in ns:
        res += elem
    return res

sum_all(5) #5, ns:()
sum_all(5, 6) #11, ns:(6,)
sum_all(5, 6 ,7 ,8, 9 , 10) #45, ns:(6, 7, 8, 9, 10)
```

```python
def greet(name, **opt):
    full_name = name
    if "title" in opt:
        full_name = opt["title"] + " " + name
    if "surname" in opt:
        full_name = full_name + " " + opt["surname"]
    return "Hi {}!".format(full_name)
print(greet("Barack")) #Hi Barack! opt: {}
print(greet("Barack", surname="Obama"))
#Hi Barack Obama! opt: {'surname': 'Obama'}
print(greet("Barack", surname="Obama", title= "Mr"))
#Hi Barack Obama! opt: {'surname': 'Obama', 'title': 'Mr’}
```
