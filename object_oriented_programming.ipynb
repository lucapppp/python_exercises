In Python every value has a type. Also we now know that Python is strongly,
dynamically typed.
- Dynamic typing: runtime objects (values) have a type, as opposed to
static typing where variables have a type
- Strong typing: the type of a value does not change in unexpected ways

In Python everything is an object
… including basic types (int, float, str,..), complex data
structures (list, tuples,…) and even functions.
What does this mean?
Every value/entity in Python may have, additional:
- Metadata: also called attributes, complementary
pieces of information
- Functionalities: also called methods, Python functions
that usually read, manipulate and modify an object’s
attributes. A method defines an ability of an object.

Three main operations can be performed with objects:
1. create the object
• of a specific type
• with its own attribute values
2. manipulate the object
• by using its methods
• directly accessing its attributes
3. destroy the object
• either by using the del command
• or by leaving it to the Python interpreter (automatic disposal):
when an object exits from every scope –thus is not reachable
by any variable– it is deleted by the interpreter.
This automatic operation is handled by a software component
named garbage collector.

A class is a pattern to create objects. An object
created from a given class C is an instance (or
object) of the class C.
A convenient way to group together:
• Data, information
• Functionalities, operations
Using classes, it is easy to:
• Write coherent software
• Reuse code
• Distribute software
• Build upon existing code
• ‘More code’ or ‘long scripts’ are not necessarily a
good thing
• Rather, ‘more functionalities’ are a measure of
goodness of the software
• To write less code – while having more
functionalities– we need to use some softwarearchitecture method: functions are one such
example
• Functions help programmers to achieve
decomposition and abstraction

Abstraction
Representing essential features while hiding
internal details and implementation
• For example, we know what the print( ) function
does, but we do not know how it works.
• It allows to use functionalities without the need of
knowing or understanding the background details.
• It is a key concept to reduce the complexity of large
systems.
• The same function can have more than one
implementation (typically the implementation of low
level functions depends on the hardware and operating
system); but this details are hidden to the user of the
function.

In programming, we can achieve abstraction and decomposition
by structuring the code accordingly.
The source code can be organized into modules, each of which:
• Implements a sub-task;
• Is self-contained;
• Is intended to be reused (a module is implemented and tested
once, but reused many times);
• Keeps the code organized.
We already presented functions as a means to improve a better
code organization. 
Classes are another helpful construct for code organization,
oriented to abstraction, decomposition and reusability.




Relevant concepts (non-strict definitions)
script: Python file intended to be run directly and achieving some
result; typically, scripts contain code written outside the scope of
any classes or functions.
module: Python executable file (.py extension) that contains
collections of functions and global. E.g., datetime, Regex, Random.
package: simple directory having collections of modules and a
__init__.py file by which the interpreter identifies it as a package. E.g.,
numpy, pandas
library: umbrella term that loosely means “a bundle of code”; a
library can have tens or even hundreds of individual
modules/packages that can provide a wide range of functionality.
E.g., matplotlib and seaborn are plotting libraries; Pytorch is a
machine learning library.

A class is a data type with additional features; in general,
it may be part of a hierarchy of classes.
A class is used as a template (or blueprint) to instantiate
objects.
In Python (and in most Object Oriented-programming
languages), each class has:
• A mandatory name (which is also the name of the data type)

• (Possibly) attributes (characteristics)
• (Possibly) methods (functionalities)
• Attributes and methods inherited from the super class.

Glossary

class: A programmer-defined type.

instance: An object that belongs to a class.

instantiation: The creation of a new object.

attribute: One of the named values associated with an object.

method: A function that is defined inside a class definition and is invoked
on instances of that class.
Two distinct methods:
-pure function: A function that does not modify any of the objects it receives as arguments. Typically, pure functions have a return statement.
-modifier: A function that changes one or more of the objects it receives as arguments. Most modifiers are void; that is, they return None.

```python
import math

class Coordinate:
    def __init__ (self, x, y):
        self.x = x
        self.y = y
    def distance(self, other):
        return math.sqrt((self.x - other.x) ** 2 + (self.y - other.y) ** 2)
    
p1 = Coordinate(1,2)
p2 = Coordinate(3,4)
distance = p1.distance(p2)
print(distance)
```

Rappresento frazioni e operazioni

```python
class Fraction:
    def __init__(self, n, d): #n for nominator and d for denominator
        if type(n) == int and type(d) == int and d != 0:
            self.nominator = n
            self.denominator = d
        else:
            print("ERROR: no numeric input")
    def addition(self, other):
        new_nominator = self.nominator * other.denominator + self.denominator * other.nominator
        new_denominator = self.denominator * other.denominator
        new_fraction = Fraction(new_nominator, new_denominator)
        return new_fraction
    def __str__(self): #str is to make it readable
        return f"{self.nominator}/{self.denominator}"
    

f1 = Fraction(3,4)
f2 = Fraction(4,5)
sum = f1.addition(f2)
print(sum)
```

Inheritance
Inheritance is a key concept in Object Oriented-Programming:
• It allows to define a class upon another class
• It creates a hierarchy among classes
• It facilitates:
    - Code reuse
    - Abstraction
    - Decomposition

Inheritance allows to define a new class
(Derived) that inherits all the attributes
and methods of an existing parent class
(Base).

The Derived class can specialize the
Base class by adding further elements
(i.e., attributes and methods).

Inheritance is modeled as an is_a
relationship: the new Derived class is_a
specialized version of the Base class.

how do we check that a class is a specialization of another??

```python
class Base:
    pass
class Derived(Base):
    pass

b = Base()
d = Derived()

print(f"b is instance of Base: {isinstance(b,Base)}")
print(f"b is instance of Derived: {isinstance(b,Derived)}")
print(f"d is instance of Base: {isinstance(d,Base)}")
print(f"d is instance of Derived: {isinstance(d,Derived)}")
```

method overriding
what if method was implemented by both person and student and they do different thing. What should i call? It reaches the closest one in hierarchy

```python
class Person:
    def __init__(self, name):
        self.name = name

    def greet(self):
        print("Hello!")

class Student(Person):
    def greet(self):
        Person.greet(self)
        super().greet() #super() means to call it the next class in the hierarchy
        print("I'm a student!")


s = Student("Bob")

s.greet()
```

a special way to call super is to use it on the constructor of the class. What if i wanna say that a student has all the charachteristic of super class and i may wanna add the class of the university.
i first use super() and then self.uni = university

```python
class Person:
    def __init__(self, name):
        self.name = name

class Student(Person):
    def __init__(self, name, university):
        super().__init__(name)
        self.uni = university


s = Student("Bob", "Polimi")
print(f"Name: {s.name}, University: {s.uni}")
```
