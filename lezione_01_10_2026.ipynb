```python
#Suppose you need to count the number of occurrences of each number in a list
l = [1, 2, 2, 2, 1, 1, 7, 7]
counter = {}
for number in l:
    if number in counter.keys(): #all'inzio è vuoto, quindi vado all'else
        counter[number] += 1
    else:
        counter[number] = 1

print(counter)
```

```python
#list of tuples
votes = [("Luca","Programming", 30), ("Luca","Databases", 20), ("Davdide","Programming", 28), ("Davide","Databases", 29)]

#dictionarie of dictionaries
votes_dict = {
    'Luca': {
        'Programming' : 30,
        'Databases': 20
    },
    'Davide': {
        'Programming': 28,
        'Databases': 29
    }
}
```

```python
#come passo da list of tuple to a dictionaries of dictionaries
votes = [("Luca","Programming", 30), ("Luca","Databases", 20), ("Davdide","Programming", 28), ("Davide","Databases", 29)]

votes_dict = {}

for name, subject, vote in votes:
    if name not in votes_dict:
        votes_dict[name] = {}
    votes_dict[name][subject] = vote

print(votes_dict)
```

enumerator
enumerator creates an order sequence of tuples, where in it i have both the value of the initial list and the index

```python
for t in enumerate(['a','b','c']):
    print(t[0],t[1])

#oppure
for t in enumerate(['a','b','c']):
    index, value = t
    print(t)

#oppure
for index,value in enumerate(['a','b','c']):
    print(index,value)
```
