## basic notions
we usually navigates true files using a graphical interface. We have a convenient hirearchy of folders.
The file systems is just what we see, they are not necessarly this way. The file system is just a sort of layer, an abstraction.
We have 3 abstraction:
File:
• we perceive the files as continuous stream of information;
• however, physical data may not even be stored close to each other
(a single file, if big enough, could be split into blocks)
File name:
• the user does not need to remember physical information about the
file (such as the disk address where it starts, if it is split in blocks,
where it ends)
• it is sufficient to assign to a file an intuitive name so to access and
manage it by name
Directory hierarchy:
• folders (or directories) are containers for files or other folders
• folders are a convenient way to keep the files organized

file handing in python
There is a schema
We retrive file location (having the name can be enough) --> we have to give the correct path.
Once we retrive it we can open or create (if there isn't a file with that name) the file.
Then we read/write the file
In the end we close the file, important because someone else could work in parallel
How to open a file?
open(filename, mode) --> filename is the path (if it's in the same folder i can just say the name). mode is the method we want to use to open the file (4 methods avaliable)
the function returns a file object that we can read or write into.
4 modes avaliable "t" --> just write, "w" --> for writing (if there's already something it replace it), "a" --> appeding content, "x" creates a file if not existing and returns an error if the files already exists.
Modes is not mandatory, there is a default which is r mode
We have an addition. We can add + after the letter and it changes the sense
The symbol ‘+’ can be added after these options:
‘r+’ opens the file for reading and writing,
‘w+ for writing and reading,
‘a+’ for writing and reading.
Additionally, it is possible to specify the type of the file:
• “t”: text file [DEFAULT]
• “b”: binary file


```python
f = open("folder/filename.txt") #r
f = open("folder/filename.txt", "w+") #read write
f = open("folder/filename.txt", "a") #append
```

once opened a file we can read it.
4 ways to read it (if i opened and stored my file in variable f):
f.read() --> reads all the file and returns a big string
f.read(n) --> reads the first n characters of the file and returns a big strings
f.readline() --> reads and returns just one line of the file. if i just opened my file it reads the first line, then it goes to the second ecc... it's like there is a cursor to see where i left. se eccedi a livello di linee mi dà stringhe vuote
f.readlines() --> reads all the lines of the file one by one ad returns them in a list of strings. It's very usefull. each time i operate i read function, even readlines, it starts from where i left, from where there's the cursor

```python
f = open("demofile.txt")
for line in f: #this means for lie in f.readline
    uppercase_line = line.upper()
    print(uppercase_line)
```

with statement
it guarantees that the file is close after the operations in the block of code, even in ther case of an error. The closing is automatical, but i have to remember to identate

```python
with open("demofile.txt", "r") as f:
    for line in f:
        print(line)
```
