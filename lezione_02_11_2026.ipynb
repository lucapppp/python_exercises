## data visualization

Data visualization is widely used in any phase of data science.
Exploratory data analysis:
• Explore unknown datasets
• Explore unknown problems
• Preliminary evaluation of results
Representation of datasets and results:
• Presentations and slideshows
• Collaborative discussions
• Publications
Why data visualization matters
Data summarization and descriptive statistics are a basic approach to
data visualization.
Data is aggregated to provide a more comprehensive of its properties
Why data visualization matters
Still, plots are a much more powerful and effective tool for representing data.
· Allow the user to dive into the data;
· Allow the user to understand trends, patterns and correlations;
· Are an tool to reason about quantitative information.
Data visualization libraries in Python
There exist many different libraries for data visualization in Python.
Features to describe / distinguish between the many libraries available for
data visualization:
Type of data: what are the data structures supported by the library. Some
support built-in Python types (list, dict, …), other only supports higher
level structures (numpy arrays, pandas data frames, …)
Type of visualizations: type of graphs/plots that are supported by the
library. For specific domain there often exist specific visualization
libraries.
Flexibility: how much the plots can be customized.
Interactivity: some libraries only allow to render plots as static images,
while others generate interactive visualization that the user can manipulate
(e.g., plotly).


Introduction to matplotlib
General minimal procedure to plot data with matplotlib:
1. Import the library
2. Call the function to create the chosen visualization
3. Render the graph (no action needed on Jupiter notebook) and/or save it
as an image

```python
'''A line plot is the most trivial
example of graph. It consists in a
series of 2D points connected by
a line.
x and y must be two lists of the same length
'''

import matplotlib.pyplot as plt

x = [1,2,3,4,5]
y = [2,3,1,1,6]

plt.plot(x, y)
```

```python
import math

xs = [0.1 * n for n in range(100)]
sinx = [math.sin(x) for x in xs]

plt.plot(xs, sinx)
```

```python
'''
More than one line can be
represented in the same plot.
A different color is automatically
chosen for each line.
'''

import math

xs = [0.1 * n for n in range(100)]
sinx = [math.sin(x) for x in xs]
cosx = [math.cos(x) for x in xs]

plt.plot(xs, sinx)
plt.plot(xs,cosx)

plt.legend()

#plt.close() 
```

```python
'''
Other good practices:
• Adding a title to the figure
• Providing a name/label/description for the axes (if you are planning to
use you figure for any publication, make sure that you name your axes…
and that they are readable!)
'''

import math

xs = [0.1 * n for n in range(100)]
sinx = [math.sin(x) for x in xs]
cosx = [math.cos(x) for x in xs]

plt.plot(xs, sinx, label = "f = sin")
plt.plot(xs,cosx, label = "f = cos")

plt.xlabel("x")
plt.ylabel("y = f(x)")
plt.title("The sin and cos functions")


plt.legend()
```

```python
'''
Font type and size can be changed using a dictionary of settings. In
matplotlib almost any component can be personalized (not always in a
trivial way).
'''

axis_font = {'fontname': 'Times New Roman', 'size': '15'}

plt.xlabel("x", **axis_font)
plt.ylabel("y = f(x)", **axis_font)
```

```python
plt.plot(xs, sinx,
         linewidth=4, linestyle=':', color= 'black', label = "f = sin")
plt.plot(xs, cosx,
         linewidth=10, linestyle='--', color= (0.9,0,0,0.7), label = "f = cos")
```

An interesting dataset
In the following examples, we use the Iris dataset
(very well-known in machine learning courses).
The dataset consists of 3 different types of iris’
(Setosa, Versicolour, and Virginica) petal and sepal
length/width, stored in a 150x4 numpy.ndarray
(similar to a list of 150 lists of 4 elements) → more
in Scientific Programming course
The rows are the different samples and the columns
are: 1) Sepal Length, 2) Sepal Width, 3) Petal
Length, and 4) Petal Width.

```python
'''
Scatter plots are used when we need to represent paired data (without any
line connecting points).
Here we plot the relation between petal length and petal width of each
flower in the dataset
'''
from sklearn import datasets

iris = datasets.load_iris()
x = list(iris.data[:, 2])
y = list(iris.data[:, 3])

plt.scatter(x,y)
```

```python
x_setosa = list(iris.data[:50,2])
y_setosa = list(iris.data[:50,3])

x_versicolor = list(iris.data[51:100,2])
y_versicolor = list(iris.data[51:100,3])

x_virginica = list(iris.data[101:150,2])
y_virginica = list(iris.data[101:150,3])


plt.scatter(x_setosa, y_setosa, label = 'setosa')
plt.scatter(x_versicolor, y_versicolor, label = 'versicolor')
plt.scatter(x_virginica, y_virginica, label = 'virginica')

plt.legend()
```

```python
'''
Boxplots provide a more compact representation of a distribution;
the distribution is represented in terms of minimum, maximum,
quartiles and outliers
'''

plt.boxplot([x_setosa, x_versicolor, x_virginica])
plt.xticks([1,2,3], ["setosa","versicolor","virginica"], rotation = 'vertical')
```

```python
'''
Violin plot are similar to box plot but shows the shape to the distribution
rather than the quartiles.
'''

plt.violinplot([x_setosa, x_versicolor, x_virginica])
plt.xticks([1,2,3], ["setosa","versicolor","virginica"], rotation = 'vertical')
```

```python
plt.figure(figsize=(10,5))
days = ["03-05","03-06","03-07","03-08","03-09","03-10","03-11","03-12","03-13","03-14","03-15","03-16","03-17","03-18","03-19","03-20"]
data = [234,245,4356,563,674,578,87967,96,876,769,769,697,796,679,7689,234234]

x = range(len(data))
plt.bar(x,data)
plt.xticks(x,days,rotation='vertical')
```

```python
'''
It is possible to force any plot (including bar plot) to use a log-scale, to
make differences among bars clearer, allowing to plot larger ranges in
the same space).
'''

plt.figure(figsize=(10,5))
days = ["03-05","03-06","03-07","03-08","03-09","03-10","03-11","03-12","03-13","03-14","03-15","03-16","03-17","03-18","03-19","03-20"]
data = [234,245,4356,563,674,578,87967,96,876,769,769,697,796,679,7689,234234]

x = range(len(data))
plt.bar(x,data)
plt.xticks(x,days,rotation='vertical')
plt.yscale("log")
```

```python
'''
Matplotlib allow to display heterogenous plots within the same panel. In this
case we plot both the bar plot and a line plot representing the moving average.
'''
moving_avg = [sum(data[max(0,i-1):min(i+2, len(data)+1)])/
              len(data[max(0,i-1):min(i+2, len(data)+1)])
              for i in range (len(data))]
plt.bar(x,data)
plt.plot(x,moving_avg, color='red', label = 'moving_avg (3)')
plt.xticks(x, days, rotation ='vertical')
plt.legend()
```

```python
'''
The histogram is designed to provide an overview of the distribution of a
series of numbers, measures, counts,…The bins allows to change the
resolution of the histogram.
'''

import random
x = []
for _ in range(1000):
    x.append(random.uniform(-10,10))
plt.hist(x, bins=1000)
```
