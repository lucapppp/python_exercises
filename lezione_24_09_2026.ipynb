primality test: remembering if a number is prime or not
write a program that takes as input a number num and tests if it is prime

```python
num = int(input("Che numero vuoi vedere sia primo? ")) #i ask the number
divisor = 2 #i start the divisor
isPrime = True #this is usefull to check is it's prime or not

while divisor < num: #i check all the numbers smaller than num
    if num % divisor == 0:
        isPrime = False #if it's a multiple, than the condition becomes false
    divisor = divisor + 1
#if i dont find a divisor, the isPrime statement keep staying true, so it's a prime number
if isPrime:
    print(f"{num} is a prime number!")
else:
    print(f"{num} isn't a prime number :(")
```

Sentinel values are used to write programs that acquire and elaborate a sequence
of values; a sentinel value indicates the end of the sequence.
Suppose we want to compute the sum of a series (of any length) of manually
inserted numbers; the sequence is terminated by the sentinel value 'N'.

ex. i wanna write a code that makes me insert number till i write N

```python
MSG = "Insert a number or N to exit" #così dichiaro una costante, che mi servirà per chiedere la frase ogni volta che voglio
total = 0.0

last_inserted = None
while last_inserted != 'N':
    last_inserted = input(MSG)
    if last_inserted != 'N': #i type this again, because saying that last_inserted = None, i enter the loop. So i wanna check again if it's a number or 'N'
        total += float(last_inserted) #+= is like typing total = total + float(last_inserted)

print(f"The sum is {total}")
```

ex. Compute the mean of a series of input numbers terminated by a sentinel value
Intuition:
1. Define a variable num_values to keep track of the (amount of) inserted values.
2. Read all numbers (until the sentinel value is inserted); add each number to the total and
increment the value of num_values.
3. If at least one number was inserted, compute the mean.

```python
MSG = "Insert a number of N to exit: "

total = 0.0
num_values = 0
last_inserted = input(MSG)

while last_inserted != 'N':
    total += float(last_inserted)
    num_values += 1
    last_inserted = input(MSG)
    
if num_values > 0 :
    values_mean = total / num_values
    print( "Mean =", values_mean)
```

there's the max function
it gives me the maximum number of a series of input numbers terminated by the sentiel value

```python
MSG = "Insert a number of N to exit: "

current_max = None
last_inserted = input(MSG)

while last_inserted != 'N':
    last_inserted = float(last_inserted)
    if current_max == None or last_inserted > current_max:
        current_max = last_inserted
        last_inserted = input(MSG)

if current_max != None:
    print( "Max =", current_max)
```

Nested loops
we combine while loops in the same way i combine if loops. i just insert the whil idented
the inner loop is executed when the outer loop is true
they might be useful with bidimensional values like matrixes or cells
ex.: implement a program that writes a 10x10 table that contains the
multiplication tables of all the numbers between 1 and 10.

```python
N = 1
while N <= 10:
    M = 1
    while M <= 10:
        print( '{:4d}'.format(N*M), end= "")
        M += 1
    print() #it prints an empty line. after each operation it goes to the next line
    N += 1
```

for loops

range(stop): creates a list of numbers from 0 to stop (excluded).
range(start, stop): creates a list of numbers from start to stop (excluded).
range(start, stop, step): creates a list of numbers from start to stop (excluded) with an
incrementation of step.

ex.:

```python
for x in range(5):
    print(x)
```

```python
for x in range(2, 5):
    print(x)
```

1. make the factorial

```python
N = int(input("Insert a number: "))
fact = 1

if N < 0:
    print("Are we wasting time or what?")
elif N == 0:
    print(1)
else:
    while N > 0:
        fact *= N
        N -= 1
    print(fact)
```

2. fibonacci sequence. Every number is the sum of the previous one, apart from the first two which are 1 and 1

```python
element_required = int(input("Tell me the number of fibonacci sequence u want to know"))
last = 1 
next_to_last = 1
pos = 3 #indice di posizione

while pos <= element_required: #<= is important to do the right number of iteration. if i stop at < element_required it'll give me n-1th number
    fib = next_to_last + last
    next_to_last = last
    last = fib
    pos += 1

print(f"The N-th number is {fib}")
```
