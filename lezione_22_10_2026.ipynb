use of any and all operations
any is a built in function that return true if any of the items given as arguments of the function is true
If every is false, then it return false, if there's at leaft one true, it returns true
syntax:
print(any(False, True, False, False)) #True

```python
list1, list2 = [], []
for i in range(1,11):
    list1.append(4 * i)
for i in range(0,10):
    list2.append(list1[i] % 5 == 0) #qui metto true nella lista se il valore in list1 è divisibile per 5
print("See whether at least one number is divisible by 5 in list 1 ==>", any((list2)))
```

all returns true if all the items are true and it returns false if at least one is false
all to an empty iterable gives true, while any gives false

```python
list1, list2 = [], []
for i in range(1,21):
    list1.append(4 * i - 3)
for i in range(0,20):
    list2.append(list1[i] % 2 == 1) #here i test if some of this number are odd
print("See whether at least one number is odd in list 1 ==>", all((list2)))
```
