## list comprehension
we write in a much compact way a thing we already know
it can be applied to set and dictionaries

```python
#up to know
my_list = []
for i in range(10):
    my_list.append(i*2)
print(my_list)
```

```python
#i can make it faster with list comprehension
my_list = [i*2 for i in range(10)] #i*2 is the operation in the second part of the for explained late. i don't need append

#i can make conditions too
my_list = [i*2 for i in range(10) if i%2 == 0] #the condition goes in the end. before making the expression it checks the condition

#i can have more conditions
my_list = [i*2 for i in range(10) if i%2 == 0 and i > 5] 
my_list
```

```python
#can i iterate on multiple collections? Yes, I can
my_list = []
listA = ['a','b']
listB = ['y','z']
for a in listA:
    for b in listB:
        my_list.append(a+b)

#posso scriverlo with list comprehension
my_list = [a+b for a in listA for b in listB] #be carefull with the order
```

```python
#output manipulation
fruits = ["apple", "cherry","banana", "kiwi"]

new_list = [x.upper() for x in fruits]
new_list2 = ['hello' for x in fruits]

new_list = [x if x != 'banana' else 'orange' for x in fruits]
new_list
```

new_dict = {expr(k):expr(v) for k,v in existing_dict.items()}

```python
#Given a sentence, write a program that saves in a dictionary all the words with their length.
sentence = input("Insert a sentence: ")
my_dict = {word:len(word) for word in sentence.split()}
my_dict
```

```python
#From the dictionary with the words and their lengths, save a list containing all the possible (unique) lengths of the words .

sentence = input("Insert a sentence: ")
my_dict = {word:len(word) for word in sentence.split()}

my_set = {values for values in my_dict.values()}

print(list(my_set))
```
