```python
#Write a program that asks the user two numbers and computes their Greatest Common Divisor
num1 = int(input("Give me the first number: "))
num2 = int(input("Give me the second number: "))
div1 = num1
check = True
while div1 > 0 and check:
    div2 = num2
    if num1 % div1 == 0:
        while div2 > 0 and check:
            if num2 % div2 == 0 and div1 == div2:
                gcd = div1
                check = False
            div2 -= 1
    div1 -= 1
print("GCD is", gcd)
```

```python
num = int(input("Tell me a number N: "))
isPrime = False

while not isPrime:
    num += 1
    isPrime = True
    for div in range(2, num):
        if num % div == 0:
            isPrime = False

print(f"The next prime number after N is {num}")
```
