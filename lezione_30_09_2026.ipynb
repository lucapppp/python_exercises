examples

```python
#Write a programm to tell if a string is palindrome
s = input("Insert a string: ")
if s == s[::-1]:
    print("Is palindrome!")
else:
    print("It's not palindrome!")
```

other way

```python
s = input("Insert a string: ")
rev = ""
for x in s:
    rev = x + rev
if s == rev:
    print("Palindrome")
else:
    print("Not palindrome")
```

other way

```python
s = "civic"
print('The word \'{}\' is {}palindrome'.format(s, '' if s == s[::-1] else 'not '))
```

```python
#How to find all the indexes of a given word?
s = "In Progr and DB bla bla bla bla and nothing."
to_find = 'and'
index = 0
new_start = 0
while index < len(s):
    s = s.lower()
    index = int(s.find(to_find))
    if index == -1:
        break
    print("One occurance was found at", index + new_start)
    index += to_find
    new_start += index
```
