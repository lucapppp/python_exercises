More in general, building a recursion means:
· Decomposing a problem in smaller instances of the same problem; we call
such decomposition recursive step;
· Such that at least one instance of the problem is so small, that can be solved
without further decomposing it; we call this base case.

```python
def factorial(n):
    if n==1 :
        return 1
    else:
        return n * factorial(n-1)
#function call
factorial(1)
```
