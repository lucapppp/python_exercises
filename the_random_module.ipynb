# random module
we won't use often external library in this course
random generates random numbers

```python
import random #import a library, then i can use its functions
random.random() #random number
random.randint(1,5) #random number from 1 to 5 included
random.randrange(5) #random number from 0 to 5
random.randrange(0,200,20) #start stop step
```

other random functions
we have random.choice --> Returns a random element from the non-empty sequence seq
random.choices --> Returns a k-sized list of elements chosen from the population with replacement weights: A list were you can weigh the possibilities for each value cum_weights: A list were you can weigh the possibilities for each value, only this time the possibility is accumulated
random.seed(a) --> Used to initialize the random number generator.
