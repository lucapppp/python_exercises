## Exercise 1
Write a Python snippet that returns its complementary sequence when given a DNA sequence as input. In DNA, the base pairs are complementary as follows:

- A (Adenine) pairs with T (Thymine)
- T (Thymine) pairs with A (Adenine)
- C (Cytosine) pairs with G (Guanine)
- G (Guanine) pairs with C (Cytosine)

Input: A string representing a DNA sequence (e.g., "ATGCATGC").

Output: A string representing the complementary DNA sequence (e.g., "TACGTACG").

```python
dna = "ACGGTTCG"
cdna = ""
for base in dna:
    if base == "A":
        cdna += "T"
    elif base == "C":
        cdna += "G"
    elif base == "G":
        cdna += "C"
    elif base == "T":
        cdna +="A"
print(cdna)
```

## Exercise 02
Write a Python snippet that calculates the GC content of a DNA sequence.

The GC content is the percentage of nucleotides in a DNA sequence that are either G (Guanine) or C (Cytosine).

Input: A string representing a DNA sequence (e.g., "AGCTATAG").

Output: A float representing the GC content as a percentage (e.g., 37.5).

```python
dna = "CGAACTGCA"

gs = dna.count("G")
cs = dna.count("C")

print(f"The percentage is {((gs+cs)/len(dna))*100}")
```

```python
#alternative
gc = 0
for base in dna:
    if base == "C" or base == "G":
        gc += 1
print(gc/len(dna)*100)
```

## Exercise 04
Write a Python snippet that takes a DNA sequence as input and returns a list of codons.
A codon is a sequence of three nucleotides. Your snippet should break the input DNA sequence into codons and return them as a list.

Input: A string representing a DNA sequence (e.g., "ATGCGTACG").

Output: A list of codons (e.g., ["ATG", "CGT", "ACG"]).

```python
dna = "ATGCGTACG"

list_of_codons = list()
codon = ""
i = 0

for base in dna:
    if i%3 == 0 and i > 0:
        list_of_codons.append(codon)
        codon = ""
    codon += base
    i += 1
print(list_of_codons)
```

```python
#oppure
list_of_codons = []

for i in range(0,len(dna),3):
    codon = dna[i:i+3]
    list_of_codons.append(codon)

print(list_of_codons)
```
